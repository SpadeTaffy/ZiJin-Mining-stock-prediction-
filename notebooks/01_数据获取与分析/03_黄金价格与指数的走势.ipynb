2.紫金矿业业务发生了变化,它的铜,锂业务同时收周期爆发影响,显著提高了它的利润,进而推动股价走高.等权重指数压低了紫金矿业的业务,增强了以黄金为主营业务的公司的比重,因此与黄金走势更加相关  
而在实际的指数编撰中,大部分股票仓位也不会超过10%,因此,分散的股票持仓构建的指数可能更接近于实际的金价走势.
